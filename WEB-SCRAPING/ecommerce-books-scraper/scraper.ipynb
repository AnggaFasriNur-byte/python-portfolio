## put everything together inside the loop

```python
title = []
price = []
stock_available = []
rating = []


for result in results :
    # title
    try:
        title.append(result.find('h3').get_text())
    except:
        title.append('n/a')

    #price
    try:
        price.append(result.find('p', {'class' : 'price_color'}).get_text())
    except:
        price.append('n/a')

    #stock_available
    try:
        stock_available.append(result.find('p', {'class' : 'instock availability'}).get_text(strip=True))
    except:
        stock_available.append('n/a')

    #rating
    try:
        rating_text = result.find('p', {'class': 'star-rating'}).get('class')[1]
        mapping = {'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5}
        rating.append(mapping.get(rating_text, 'n/a'))
    except:
        rating.append('n/a')
```

```python
book_scrape = pd.DataFrame({'Title' : title, 'Price' : price, 'Stock_Available' : stock_available, 'Rating' : rating})
```

## pagination

```python
title = []
price = []
stock_available = []
rating = []

for i in range (1, 51):
    website = 'https://books.toscrape.com/catalogue/page-' +str(i) + '.html'

    response = requests.get(website)

    soup = BeautifulSoup(response.content, 'html.parser')

    results = soup.find_all('article',{'class' : 'product_pod'})
    
    for result in results :
        # title
        try:
            title.append(result.find('h3').get_text())
        except:
            title.append('n/a')

        #price
        try:
            price.append(result.find('p', {'class' : 'price_color'}).get_text())
        except:
            price.append('n/a')

        #stock_available
        try:
            stock_available.append(result.find('p', {'class' : 'instock availability'}).get_text(strip=True))
        except:
            stock_available.append('n/a')

        #rating
        try:
            rating_text = result.find('p', {'class': 'star-rating'}).get('class')[1]
            mapping = {'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5}
            rating.append(mapping.get(rating_text, 'n/a'))
        except:
            rating.append('n/a')
```

```python
book_scrape = pd.DataFrame({'Title' : title, 'Price' : price, 'Stock_Available' : stock_available, 'Rating' : rating})
```
